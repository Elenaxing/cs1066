**There's no single approach to working with an AI assistant.**

*   For coding, we'll lean toward a single approach: AI always generates the Python code.
*   For design, you're free to explore the full range of options.

1.  **Move fast.** I know some people who have an idea and immediately write a prompt.

    *  They want to see what AI can do.
    *  They often don't care about grammar and punctuation in writing the prompt.
    *  BENEFITS: Fast way to get started, and the result might be good and work. You'll learn a lot about a model's capabilities.
    *  COSTS: The resulting program might not work or be close to what you imagined. You'd better be good at debugging AI-generated code.

EXAMPLE: I structured the last pair-share activity to put you in this mode.

2.  **Design first.** At the other end of the spectrum are people who take design as a serious task that needs to be completed before coding.

    *  They still use AI, but they get the AI to help think through the design decisions.
    *  They know that clarity in a prompt can produce better outputs, and they enlist AI help for writing a clear prompt.
    *  They often decompose the problem into pieces and methodically work through the design of each piece as well as the wiring of the pieces together.
    *  BENEFITS: Slower to first prototype, but the code is more likely to work and need few major changes. You understand the hard parts of the problem from the design stage, and you can ask the AI about how its code handles those challenges.
    *  COSTS: More sunk costs if you decide to throw your first prototype (but not the learnings) away.

Example: In the IDE's editor, open `ai_transcripts/Prompt-edgar_financial_trends_dashboard.md` and preview it. Take a minute to skim through it.

Both (1) and (2) present the AI with a prompt that asks it to create a complete solution. There is a third approach.

3.  **One piece at a time.** You decide on a decomposition and then follow (1) or (2) for each piece of the problem.

    *  This allows you to learn and experiment along the way.
    *  The previous approaches work when you know what you want.

### Start Small and Simple

When you're not entirely sure where you're going and what you need, it pays to start small and simple.

> This adage applies to working with AI as well as computational problem solving in general.

We followed this adage with the prompt that built `m01/step1.py`, but this prompt also worked for two important reasons:

*   It fell within **the AI's knowledge frontier**, which is a concept that we will discuss later.
*   It fell within **my knowledge base**.

This combination is **the sweet spot** for which you should shoot.

Why?

1.  By starting small and simple, it's easy to check what the AI has generated. "Easy to check" because it is in your knowledge base.

2.  Given something that works correctly, you next ask the AI to build on it.

3.  And now, a small step forward is a lot easier to check than a large and complex one.

4.  Return to (2) and iterate until complete.

### Back Again to Thinking About Process

Our problem-solving process has become:

1.  Identify a task you want automated
2.  Set the requirements for the tool that'll automate the task
3.  While the tool is incomplete:

    3a.  Pick a(nother) small part and give its requirements to the AI

    3b.  Check what the AI built

4.  Use the tool and focus on the interesting work

In general, (2) is writing a **problem specification**. It specifies the problem's important qualities and requirements.

**QUESTION:** Do you see these qualities and requirements in our original prompt?

> **PROMPT:** Write a Python program called m01/step1.py. This program should fetch AMD's company data from the SEC EDGAR API. The CIK number for AMD is 0000002488. The endpoint is https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json. Include a User-Agent header, and save the returned JSON in a file called amd_data.json. Use the standard library.

And in general, the loop in (3) depends upon **problem decomposition**, which is the breaking of a problem into a bunch of small tasks.

**BEST PRACTICE:** It typically doesn't matter on which decomposed task you start. Possible choices include:

*   **A critical task.** Experts know that such tasks should consume more of their focus, because they're complex or the design is murky. They're important to get right to the success of the entire project.
*   **An easy to solve task.** Experts might start with an easy task to get a quick win under their belt.
*   **One previously solved.** Experts might reuse code from a previously completed project, which again moves them past the blank canvas.

### Specifications vs. Implementations

**A specification** says **what you want done**. It’s the goal.

**An implementation** says **how to do it**. It’s the method.

You dictate the **what**, and the AI generates the **how**.

Should you care about the **how** if it's done by AI?

> In working with AI, you can sometimes just specify the what. But when the AI gets confused or produces something that doesn’t quite work, that’s when your understanding of the how becomes valuable.
>
> It's why this course wants you to **think like a programmer**, who knows how to break a problem into small, clear actions. It is also why I'll ask you to learn enough to read the AI-generated code. You'll get to your desired tool faster if you can **recognize the actions you want done in the AI-generated code**.

### The main function

This leaves the big block of text starting at line 196 through line 214, which is a **function** called `main`. 

> The word "function" has a technical meaning in computer programming, which is similar to the concept of a function that you probably learned in high-school math or when using the `sum` function to add a column of numbers in a spreadsheet. It is a shorthand name for a bunch of work you want done but don’t want to see. You just want to call the function, possibly passing it some input data, and get back a result.
>
> `def` on line 196 stands for “define.” The lines indented below line 196 are bundled together into an object called `main`. They are the implementation of `main`.

Lines 17, 50, 80, and 128 define four other functions whose implementations we just hid.

> This idea of hiding the details of how a function takes its inputs and produces its outputs is called **encapsulation**. It is a powerful idea that supports **modularity** and **reusability**, which help us to build complex applications that work.

When generating code with AI, we leave the details---the bundle of statements encapsulated within a function name---to the AI.

----

### An Everyday Example of Encapsulation

Imagine that your child or a friend’s child is eager to help you cook breakfast. You want to make eggs, but you know that the child has never cooked anything in their young life.

You wouldn’t turn to them and say, “Please make scrambled eggs.” Back to `scrape.py`, you can think of `main` as "make scrambled eggs."

To a child, you can’t even say, “Grab the eggs from the refrigerator, scramble them, and put the cooked eggs on a plate.” This is like the list of commands in `main`.

You have to be far more detailed and say, “Grab the carton of eggs from the refrigerator. Get a clean bowl. Take one egg and break it, putting its contents into the bowl. Do the same with a second egg, ....” And so forth.

When we prompt an AI to translate a natural language prompt (e.g., "please make scrambled eggs") into a Python program, we expect it to generate the far more detailed instructions. The resulting statements are simple directives that the child/machine can follow.

The functions on lines 17, 50, 80, and 128 are subtasks like "break an egg" that also require more detailed instructions. This is just more decomposition until we have what can be stated to a child (or in a single Python statement).

### Verification While Staying Above the Details

You've learned a tiny bit about programs: They can be split into functions that encapsulate work. This is the first line of defense against detail-overload.

We're focusing on the `main` function because it describes the major steps (or subtasks) in our prompt. It's the place where we dive into the topmost details. And importantly, it's where we want to look to verify that the AI understood what we wanted. 

To get the gist, we don't want to focus on every detail. These things can help.

*   **Comments.** Learn to identify them. Read them! Everything not a comment is a **statement** that does something.
*   Understand that comments often precede **a block of statements** that perform the work.
*   The `=` (equal sign) is **the Python assignment operator**. To the left of it is **a name for a value**. To the right is **a value** (or an expression producing a value). `=` is not equality.
*   Spaces in names and **string values**.
*    **Function invocations**. Function inputs (**its parameters**) are in parentheses. Python functions **return a value**.
*   Function implementations often start with a **docstring**.

### Run a Web Scraper

I'll run `scrape.py`. I say "I" instead of "we" because this program is configured to run locally using my machine's web browser.

> If you know about `requirements.txt` and Python virtual environments, which we'll cover in section this week, you can run `scrape.py`. If not, don't worry. You'll run a version of `scrape.py` that uses a headless browser in section.

**A Throw-away Environment.** Despite Codespaces being a virtual environment, I am going to create a new virtual environment, which you should think of as a throw-away environment. Why?

-   I want to try something, but I don't want the changes I need to make to my machine's system software (or in your case, a codespace's system software) to persist beyond this experiment.
-   Remember, your CS1066 codespace is long lived: for the semester, if you don't delete it.

A **Python virtual environment** is essentially a copy of the system software put in a local directory, where I can make changes, and some changes to my environment variables, which indicate where the terminal's shell program should find programs that implement the commands I type.

1.  To create a Python virtual environment, you type `python -m venv venv`, where the first `venv` is the Python module that does the work and the second is the name I gave to my new virtual environment. It basically copies a lot of system files into a new directory called `venv`.

2.  Then you switch from the environment you don't want changed and into this new virtual environment with this command: `source venv/bin/activate`. It's basically just changing a few environment variables to point things at the `venv` directory. Notice that the terminal prompt changes.

3.  Now, in this throw-away environment, I can install new modules with the command `pip install -r requirements.txt` (pip stands for "Pip Installs Packages"). This loads a bunch of modules needed for web scraping. Since we won't be doing much web scraping beyond this week, I temporarily want these modules.

4.  Finally, I can run the scraper: `python3 scrape.py`

Watch carefully as `scrape.py`:

*   launches a browser
*   connects to Google Trends
*   interacts with the Google Trends page

It automatically clicks through the pagination buttons, gathering data from every page of results. It’s essentially emulating a human clicking the "Next" button.

When it is done, it displays the state-by-state trend data, just like we wanted!

> This illustrates the power of what you’ll learn to do.

Through a prompt, you direct an AI to automatically build a useful tool. And this tool automates some work that would’ve taken you ages to perform by hand. Without having to write any code yourself, you create a program that does necessary-but-uninteresting work, and it does it in an instant. You are now free to focus your energies on more meaningful tasks.
