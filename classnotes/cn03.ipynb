### Basic Operation of Generative AI

If you're going to rely on an AI coding assistant, you must understand:

*   how it operates, and
*   what it is not.

> If you need a basic introduction, I highly recommend [Stephen Wolfram's "What is ChatGPT Doing ... and Why Does It Work?"](https://writings.stephenwolfram.com/2023/02/what-is-chatgpt-doing-and-why-does-it-work/). It was written in early 2023, but it does a fantastic job of helping you understand the basics.

In short, **a generative AI model** takes a prompt and responds with a _token_ (i.e., a word, part of a word, or a punctuation mark) that is a _reasonable_ continuation of it.

> EXAMPLE PROMPT: The best thing about AI is its ability to

*   What is a reasonable continuation in this example?

*   Why are these answers reasonable?

**An AI chatbot** adds the output token to the end of the prompt and calls the model again. This repeats until a _stopping condition_ is met.

*   What is a stopping condition?

**An AI coding assistant** is an AI chatbot trained on mountains of code.
