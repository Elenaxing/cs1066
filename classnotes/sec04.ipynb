### Conditional Execution

If-statements allow programs to conditionally execute a block of statements. Line 20 in `cards.py` is a simple if-statement. It checks the value of a boolean variable `mixup`, and if that value is `True`, the statements indented under the if-statement are executed. If that value is `False`, the statements are skipped.

In general, the condition of an if-statement (i.e., the characters between the keyword `if` and the ending colon) can be any expression, including ones that use the logical operators `and`, `or`, and `not` to combine multiple expressions and express a complex condition.

> See if you can state in English what the condition in the if-statement on line 8 in `deck.py` does. If you need help, ask the IDE's AI to explain the line.

Notice also that this if-statement isn't just protecting the execution of the statement indented underneath it (i.e., line 9 should execute only when the condition on line 8 is `True`). This if-statement is paired with an else-statement (on line 10). Together, this code construct says the program will execute either the code under the if-statement or that under the else-statement depending on the value of the condition.

> With an if-else-statement, the program is making a choice.

Python (as do most programming languages) allows you to not only choose between two blocks of code, but create a list of conditions that determine what an if-elif-else block does. Here's a simple example that maps a numerical score to a letter grade:

````python
score = 79

if score < 60:
    print ("Your grade: E")
elif score < 70:
    print ("Your grade: D")
elif score < 80:
    print ("Your grade: C")
elif score < 90:
    print ("Your grade: B")
else:
    print ("Your grade: A")
````

The numerical score is converted by the if-elif-else block into one of five letter grades.

That's the basic mechanism for controlling which statements in a program are executed when the program is run. If you have any questions about if-statements, their many forms, or how their conditions are expressed, please ask. We'll assume you understand the gist of what's going on with them from this point forward.

### Looping

Once you understand conditional execution, you're ready to understand loops, which are blocks of statements that are run multiple times (including 0 times). Python has two basic forms of loops and a number of syntax shortcuts that embed loops in the construction of objects like lists.

Line 12 in `cards.py` is one example. It is what is called a _list comprehension_. The technical term isn't important. We'll simply use this shorthand to introduce you to the general loop syntax and operation.

> Open `cards.py` in your IDE's editor. Highlight line 12, and prompt the AI to "Please write this statement using a for-loop rather than a list comprehension."

It should write code that first sets `ranks` to an empty list, and then executes a for-loop that appends each number `n` in the range between 2-11 as a string to the end of the list `ranks`.

-   The indented statements under the for-statement are the loop's body. It is executed some number of times defined by the length of the _iterable_, which is a data collection (e.g., the characters in a string or elements in a list) or a sequence, which is what `range` produces.
-   In order, the elements of the iterable are given the name `n` (i.e., this loop's _loop variable_) and an instance of the loop body executes.
-   When every element has had a chance as the name `n` in the loop body, the loop's execution terminates (i.e., the statements after this loop execute).
-   You use a for-loop when you want a loop that executes a known number of times. In this example, the loop executes 9 times, which is the count of numbers between 2 and 10 inclusive. In Python, by definition, the first number in the range is included and the last is not.

Lots of details. Just remember that this is a specific kind of loop.

### Two Most Common Situations

`git` is a powerful system, and there's a lot that you can do with it. We'll focus on just two common situations:

1.  You started to make some changes on a file, and you want to scrap all your recent changes and start again.

2.  You had an idea, and you've been working on it for several hours/days. Along the way, you committed the intermediate work a few times, but now you realize that the idea is doomed to fail. You'd like to revert back to the version before you started on the idea.

Both these are simple to do with git, **if you made regular commits at thoughtful points in your work**. We'll focus on the actual git commands that get you back to the state of your files that you want. We can't easily teach you to take "regular commits at thoughtful points in your work." You'll have to work out your own strategy. Everyone's is a bit different.

### Situation #1

This is the simplest situation. You just made some edits, and you want to throw away all of them. You want to revert to your last backed up file, which is at the `HEAD` of the saved versions in your local repository (i.e., in your codespace's filesystem).

In the terminal, you run this command:

```bash
git restore test.py
```

The editor should show the version that starts `# Another modification`. Your last edits are gone.

----

### Situation #2

Let's say instead of the last version, you actually want the version that starts with the comment `# First modification`.

Running `git restore test.py` won't do anything because your current working file matches the one at the `HEAD` of your backups. Instead, you have to specify which backup you want, which you do by mentioning its **commit hash**.

> Ask an AI to explain what a git commit hash is.

How do you know the hash of a commit? It was printed at the start of the message that the original `git commit` command wrote, but you probably didn't remember or didn't log it. Nobody does.

> Instead, you can ask git to print out a log of your commits with the `git log` command. Try it now.

Notice that this output is paged, and it includes every commit you made in your repository. It's not just the commits that included the file `test.py`. Often this is what you want because you committed a bunch of files that correspond to a milestone. 

> In order to exit paged output, type `q` at the colon prompt (`:`). This will return you to the default terminal prompt.

If you were really interested in only one file, you could run:

```bash
git log -- test.py
```

If you add the flag `--oneline`, the command will print just a shortened version of the hash and the commit message.

```bash
git log --oneline -- test.py
```

We want the 7 hex digits before the commit message that reads "First update." Let's say that this hash is `1234abc`. With that information, you can now run:

```bash
git restore --source 1234abc -- test.py
```

With the file reverted, you'd think you're done. You're not. You probably also want to make this version of `test.py` the HEAD of your backups. So, you'd add and commit this reverted file:

```bash
git add test.py
git commit -m "Revert test.py to version from 1234abc"
```
