### Structured and Unstructured Data

A consequential challenge you'll face involves finding the data you need on the SEC EDGAR site. Here are some hints that will help:

1.  You want the indicated company’s **10-K annual reports**. Each company uploads many different kinds of reports to EDGAR.

2.  In each annual report, you want to extract the company's values for reported revenue, net income, and total assets. The key difficulty here is that the PDF of a company's annual report is an example of **unstructured data**.

In unstructured data, it's hard to programmatically describe a consistent and concise process for finding a particular number. Think about the different ways that companies include their financial results amongst an annual report's figures and narrative paragraphs.

But the financial data in a company’s annual report comes from its accounting system, which is an example of **structured data**. For each record in the accounting system, you know exactly what it represents.

JSON is a text-based file format for storing structured data, and it is **why you should request that EDGAR respond with a company's reports in JSON**. This format makes programmatic access possible.

### A Real Example

While better, finding the numbers you want even in this programmatic manner still isn't easy. You're now challenged by the size and complexity of a JSON file that describes the finances of a real-world company.

Open `m03/amd_data.json` in the IDE's editor. This file is an example of what you'll soon download from EDGAR.

-   In the key-value pairs of the file's top-level dictionary, you'll find some basic company information, i.e., the company's CIK and its official name.
-   The hierarchy goes deep fast.
-   The data at a particular level can be quite long.

----

### Important Keys

Keys are how you can find the financial values you need.

1.  The first important key is `"facts"` (line 4). The data you want sits in this dictionary.

2.  Search for the key `"NetIncomeLoss"`, where the search term should include the opening and closing double quotes. This is the key for a company’s net income, which is a profit or loss depending on whether the company made or bled money over the identified period.


 
3.  The value for the key `"NetIncomeLoss"` is a dictionary, which contains a lot of information. You're interested in the `"USD"` key, which is another 2 nested layers deep.

4.  The key `"USD"` has a value that starts with a square bracket `[`, which means it is **a list of values**, where each value happens to be a dictionary. The U.S. dollar value for `"NetIncomeLoss"` that you _might_ want is keyed under `"val"` in these dictionaries.

5.  The other keys in these dictionaries are the context for a net income value. For example, the first `val` comes from a 10-K form that reports on the company's activity between December 27, 2009 and December 25, 2010.

----

### The Challenge

There are over 100 entries in the list that's the value of the key `"USD"`. This is because a company like AMD files quarterly with 10-Q forms and annually with 10-K forms. And they sometimes amend filings with a 10-K/A.

> Think about how you can use this knowledge as you write your prompts. What will AI encounter as it tries to extract the data you want? How do you make it clear what it should do? How do you check what it did?

While this example deals with the JSON files that EDGAR produces, this lesson applies to many other real-world APIs out there: weather data, financial markets, social media, government databases, and so forth. Most require some type of authentication and respond with highly complex files of structured data.
