### Installing Modules

You cannot import a module into your program if the module doesn't exist on your computer. This is the error you encountered when trying to run `trends_scraper.py`. The Python interpreter went to import the module, and it didn't find that module anywhere on your computing system.

Keeping track of the libraries that a program you want to run depends on, and what other libraries those libraries depend on, quickly becomes a nightmare. So programs often ship with a `requirements.txt` file, which is a text file that says what libraries and their versions the program developer used. No guessing. Just install what that developer installed.

### String values

Feel free to check your paraphrasing with a fellow student or member of the instructional staff.

The value `"vibe coding"` is what's called a **string literal**. It is a piece of text that we want to create, save, and possibly manipulate in our program. The word 'vibe' plus a space character plus the word 'coding' together comprise the string value. 

We know it is a string because Python requires you---the programmer---to wrap string values in a matching pair of open and close quote characters.  Those are called **delimiters**.  They mark the start and end of a string's value, and they are NOT part of the string. 

Python allows the matching pair of delimiters to be:

*   a single-quote character `'`, which is straight, not curly. A curly quote is a different character, and it is not a string delimiter in Python.

*   a double-quote character `"`. Again, a straight double-quote character.

*   3 single-quote characters `'''`. All straight.

*   3 double-quote characters `"""`, which you've seen used in docstrings.

Why all these options?

*   With different options, you can more easily include a single- or double-quote character in a string. In other words, if you want to use a quote in the string, you use the other choice as the string delimiter.

*   The triple-quote options allow you to represent the action of you hitting the enter key. As an example, this allows you to build a string that has 2 lines of text.
