**Part 2.** Think about how you'd prompt AI to help you create a simple [Mad Libs game](https://en.wikipedia.org/wiki/Mad_Libs) that pulled its stories from a JSON file. Here's an example JSON file:

```json
{
  "title": "A Day at the Park",
  "template": "Today I went to the park and saw a [0] [1]. It started to [2] very fast, which made me feel [3]!",
  "inputs": {
    "0": "adjective",
    "1": "noun",
    "2": "verb",
    "3": "adjective"
  }
}
```

-   You may change the JSON structure, if you wish, or use it as is. Note that the bracketed numbers in the template correspond to the parts of speech in the "inputs" dictionary. Only the numbers in the square brackets, which are expressed as strings, are the keys to the "inputs" dictionary.

-   You can use more speech types than shown here. For example, the Mad Lib may request a singular or plural noun. It might ask for a number, or a verb ending in -ing. Be creative!

Your task is to design and implement two programs:

*   The first program presents a randomly chosen Mad Lib story from a JSON file, prompts the player for a list of words that match the characteristics of the "inputs," and then prints the resulting story to the terminal (or a file). The inserted words should be highlighted in some manner.

*   The second program allows a human designer to easily put new stories into the JSON file. The designer (not the AI) must provide the story and decide what words will become blanks. The rest of the design is up to you. Make it easy and fun to add stories to the JSON file.
