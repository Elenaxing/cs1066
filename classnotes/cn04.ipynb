### Changing to a Different Interface

Recap:

*   We identified a problem that was a **tedious and time-consuming task when done manually** (i.e., collecting Google Trends data and plotting those data for easy consumption).
*   We **enumerated the steps in our manual work**, and then worked with an AI coding assistant to build the tool, which was at its core a web scraping program.

The tool worked, but it was slow and intermittently didn't work. How did we end up here?

*   We pursued a solution that directly reflected the work you---or a human assistant---would do. That is, the solution was a computer program that clicked and scraped just like a human.
*   But the tool doesn't need to emulate a human. You care only that this tool quickly gathers, analyzes, and plots data from the web.

> KEY POINT: A tool should do **what you need**. It doesn't have to do it in the same manner as you. It should do the task accurately and quickly.

The answer for quickly and robustly grabbing data off the web is a new interface. The designers of the web recognized this, and today's web has:

*   One interface that's easy for humans to use
*   Another that's appropriate for computer programs

The machine-optimized interface allows your computer to directly ask another computer for data. It avoids the other computer burying the data you want in the code that describes a human-readable webpage.

----

### Application Programming Interfaces (APIs)

A machine-optimized interface is technically called an **application programming interface**, or **API**. A website may define an API, which is a set of rules that describe how a program running on your machine can communicate with a program running a website's servers. 

The machine running a website is called a **server**. When your machine reaches out to a server, it becomes a **client** of that server. In general, computer scientists talk about servers as controlling some **resource** that clients want to access or update.

----

### Wikipedia Example

Let’s explore these ideas and use a real API. We'll use Wikipedia, which has both interfaces.

> **Silly Challenge:** On the Wikipedia article titled "Dog", grab the text of the introduction.

Using your favorite browser, go to the URL: https://en.wikipedia.org/wiki/Dog.

We (humans) know that we want the paragraphs that begin with the sentence: “The dog (Canis familiaris or Canis lupus familiaris) is a domesticated descendant of the gray wolf.”

**Understanding the challenge using Wikipedia's API**

In the **client-server programming model** mentioned earlier, our computers are clients of the Wikipedia server, which controls the text of the articles on Wikipedia (i.e., the resource).

Again, using your favorite browser:

Is it easier to find the text of the first paragraph?

### Understanding the URL That Connected Us to the API

1.  What is `https`?

    This specifies the "language" we're using to initiate communication between our machines and the website's server. Two machines on the internet can’t communicate unless they agree on the language they'll use during the communication. This is just like two people who can’t communicate unless they agree on the language they’re using. 

    For example, what I mean when I say the word "burro" depends on whether you and I are communicating in Spanish, where burro means donkey, or in Italian, where burro means butter.

    The language specified is HTTPS, which is an acronym for **Hypertext Transfer Protocol Secure**. HTTPS is a text-based conversation hidden under a layer of encryption to protect the communication's plaintext contents. The details aren't important to what you want to accomplish except to say that once you've typed this prefix, the server now knows how to interpret the meaning of what follows the `//:` characters in the URL.

2.  What is `en.wikipedia.org`?

    This is the server's **hostname**. The hostname is specified as an **Internet Domain Name**, which is an addressing standard telling the internet how to find the machine we want. Again, we'll gloss over a lot of details. We'll assume that we’re now talking directly to the server that responds to this name.

3.  What is `w/api.php`?

    Somewhere on the Wikipedia server machine is the program that implements Wikipedia's API, and this is us telling the server that we want to talk to that program about the resource that it controls.

4.  What is `?action=query&prop=extracts&exintro=1&explaintext=1&titles=Dog&format=json`?

    This is us telling Wikipedia’s API what data we want. It is called a **query string**, which is a list of **name-value pairs**.
    
    -   The question mark character (`?`) starts this query string.
    -   The name-value pairs are each separated by an ampersand (`&`) character.

----

### What Does the Query String Say?

Just as you saw in Python, a name is associated with a value using an equal sign (`=`).

*   The first name-value pair is `action=query`, which says that the action we want the server to take is to look up something. 
*   What we want to look up is the article titled "Dog", as specified in the `titles=Dog` name-value pair.
*   Three of the pairs say we want to extract the first paragraph in the introduction.
*   The last name-value pair says we want the data returned in JSON, which is another communication standard.

----

### JavaScript Object Notation (JSON)

JSON is a standard way to exchange data between programs using **structured text**, which we'll later compare against **unstructured text**. 

> As an example, look at what's in the browser window when we used the second of our two URLs.

JSON also uses key-value pairs, and it uses hierarchy to express the structure of complicated data.

### Authentication

To get access to most of the data on the web, you need to understand how a program can authenticate itself---using your credentials---to a server.

> In general, **authentication** is a process by which you present some information to a server that says who you are. A server typically uses that information to determine if you’re authorized to access the server's resources and perform whatever actions you've requested it do.

There are many reasons why web servers require authentication, including:

- Access control: An organization's server might contain data that only people with certain roles or privileges can access.

- Billing and tracking: An organization might make a server available and charge individuals by the quantity and types of access. Even if a server's API access is "free," the owning organization might track usage to analyze which types of users use which types of features.

- Legal and compliance considerations: With access control and tracking, an organization can demonstrate that their service adheres to any regulations that might apply to the server's data.

- Rate limiting and abuse control: Again, with access control and tracking in place, an organization can keep a single user from monopolizing a server's resources. And in the case of a malicious user who monopolizes a server's resources to purposefully deny the service to others, an organization can quickly revoke that user's access and restore the service.

You can read more about Wikipedia's approach to **rate limits** and how a client's identity affects these limits at [this page](https://www.mediawiki.org/wiki/Wikimedia_APIs/Rate_limits).

----

### How to Authenticate

For Wikipedia, a simple way to authenticate is to include a standard **user-agent header** in the API request.

```python
headers = {"User-Agent": "Harvard CS1066 bot"}
```

`headers` is a name, and the Python object assigned to this name is a **dictionary**. This is a data type that is built-in to Python, like integers and strings. You use a dictionary when you want to store information that is naturally expressed as key-value pairs.

> Example: The string `"User-Agent"` is a key, and the value of this key in the dictionary object named `headers` is the string `"Harvard CS1066 bot"`.

### Header vs. Content, a Real-World Example

As a real-world example of the difference between the header and the content (in our case, the content is a request), think of an email. When you want to write to someone, you open an email program and start typing “dear so-and-so ....” That’s the content, which might be a request. The email address you put in the To-field and the date and time you sent the message are header information.
