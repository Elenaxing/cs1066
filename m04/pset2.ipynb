### What I Want

This course began with an example what-I-want desire:

> An investment idea had popped into your head, and to check if it made sense, you needed to explore the annual revenue, net income, and total asset values of numerous publicly traded companies.

You knew that the SEC online EDGAR database contained the companies’ raw data, but you also knew that collecting and analyzing these data by hand would take forever. **You needed a tool** that would allow you to input a company’s name so that it would output, a second or two later, a dashboard with your three metrics pulled from EDGAR and plotted over the past several years.

We showed you such a tool, and now you'll build your own version of it.

In preparing you for this task, we've taken you through a problem-solving process that will help you tackle _any problem_ that looks like this what-I-want request. Let's quickly review its main components.

----

**1: Specify the problem-to-be-solved.** Success in problem solving always starts with a specification, which contains:

-   a clear statement of your problem
-   the inputs from which you start the computation
-   the output you desire
-   any important constraints on the solution

You practiced writing specifications with each program we've built. You know the specification for this pset problem.

In general, writing a straightforward specification helps clarify in your mind exactly what you need done, and it will help you to communicate your need to an AI. In crafting a specification, it often helps to imagine example inputs and outputs, because in generating these examples, you can check to see if you forgot some important constraint or specification detail.

Don't shortchange this step. Clarity and sharing as many details as possible---both what's necessary and what's preferred---will help keep the AI focused on your task. Leave whitespace in your problem specification where you are happy to see the AI fill in what is typical or common. We took this whitespace approach in our web app: we wanted a familiar-feeling and pleasant-looking webpage, and we left the interpretation of what this meant to the AI.

----

**2: Decompose into small tasks.** When first viewed, your problem-to-be-solved can feel intimidating. You may feel that you don't know where to start. But you can overcome this feeling by tackling the problem as a series of small tasks.

We repeatedly decomposed every problem we tackled.

-    You can express that decomposition in the prompt you give to the AI, effectively providing the AI with all the pieces it needs to build. You saw this in Aadity's one long prompt that produced our solution.
-    Or you can use that decomposition as a roadmap for multiple AI prompts, as we did in teaching you the different technologies involved in our solution.

Both these approaches work, and which you choose depends on whether you like to see:

-   Something working quickly (i.e., a small piece of the problem checked off), or
-   How much an AI can do on its own (and you're not intimidated by the large code base it will produce).

----

**3: Translate into code.** Decomposition provides small and often well-specified problems that an AI can easily and successfully translate into working programs. While you can give a big problem to an AI, the chances that it will successfully produce a working and robust solution diminish as the complexity and/or unique aspects of your prompt scale up. Today's AIs are capable of amazing feats, but the ambiguity of English (or any human language) is an impediment to describing what you want done in a reasonable number of words.

Also, to give time to grow in your own skills, you can decompose a problem into small, well-specified pieces and have the AI produce programs for these individual pieces. Then, ask the AI to bundle these programs together into a single tool that solves your problem, as you did in the last pset.

----

**4: Verify the solution's correctness.** The last step in the problem-solving process is to run each produced program and, assuming they run, verify that they act as you specified. We've started to give you the tools you need for this step.

In particular, we encouraged you to develop the skill of reading Python code for its gist. Testing is a notoriously difficult problem because no single test can cover all the situations under which any program of reasonable complexity should work. It's easier to skim the code than it is to build a comprehensive test suite, even with the help of AI. Or stated another way, it is easier for AI to help you understand the generated code than it is for AI to build you a test suite for that code.

Going forward, we'll cover other useful testing and diagnosis skills.

### So Build It

**Task:** Build a financial analysis tool that prompts its user for a company's CIK number---that's a unique identifier that the SEC assigns to each public company---and produces a webpage displaying plots of that company's revenue, net income, and assets over the past 10 years (when that many years of data exist).

We demonstrated with AMD's actual historical financial results. AMD's CIK number is 0000002488; EDGAR's interface requires the CIK's full 10 digits and so **the leading zeros are important**. The generated financial dashboard for AMD should stack the three plots.

> **CRITICAL PIECE:** We left a lot of whitespace in this task description. This is to give you agency. Don't give all this whitespace to the AI. Decide what you'd like to try and what you'd like to see done. Remember that we're excited to see what you and the AI produce, but it's not only about the final product. It's about building skills in YOU.

> **BE CAREFUL!** Make sure that you include in your prompts that you want the AI to work in the `m04` directory. If you want it to consult files that we used or built in class, copy those files into `m04`. We'll talk more about the AI's context. We're trying to balance the benefit of having all your CS1066 files under one codespace while constraining the context that the AI uses.

> **AUTHENTICATION:** The SEC EDGAR API uses fairly simple authentication. It requires you to set `"User-Agent"` in the header to a Python string of the form `"Name email_address"`. While you could use a name and email address that fully identifies you, partial identification is sufficient, e.g., `"UniversityStudent your.email@university.edu"`. Yes, not your name and email, but exactly that string. Or change "university.edu" to "harvard.edu". Your choice.
