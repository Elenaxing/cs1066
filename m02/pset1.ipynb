**CHALLENGE:** Change `trends_save.py` so that it prompts the user for a search term or phrase.

> NOTE: Do **NOT** copy this subtask challenge and give it in its entirety to AI. Yes, we know that you can do this and get good results. We'll try that in the next pset. The following instructions are **for YOU**, not an AI. Use an AI only when it says to chat with one.

Follow these (strongly suggested) steps:

    > HINTS: Write a set of straightforward statements that are easy for the AI to follow. When you think you're done writing, re-read your prompt. Is it possible to make it clearer or less ambiguous? If so, do so.

4.  Open `my_log.md` in the IDE's editor, and copy what you've written in the chat box where it says "INSERT-PROMPT-TEXT." Now you can send the prompt to the AI.

    > HINTS: Look once through all the changes before taking a second pass over them to decide whether to accept any or all.

### New Task: Improve the Tool

**CHALLENGE:** Think about the use case for this tool, and imagine ways that it could be improved so that the user spends even more time focused on their question and less time having the tool dictate the repeated work.

Here is one idea that aligns with this challenge:

> If the user wants to run the tool for multiple different search terms, the tool will overwrite the plot file for the previous search term with the new one. If the user wants to compare multiple plots simultaneously, this requires the user to rename the plot file after each run. Not very exciting work.
> 
> Could you change the tool's operation so that it automatically picks a reasonably thoughtful name for the plot file without having to query the user? The tool should tell the user the name of the output plot file, and it should only interrupt the user's workflow if a file with that name already exists in the directory.

In `my_log.md`, we've left space for you to write your own idea about how you might improve the current tool.

*   This description shouldn't be much longer than the example idea above.

After you've written your own idea about how to improve the current tool, choose either your idea or the example one, and implement it. In `my_log.md`, document your choice and then the work you did to create the improved tool.
